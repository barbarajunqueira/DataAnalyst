Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:

La funció ha de classificar el resultat en les seves respectives categories

```python
def pedir_datos():
    """
    espera del usuario 2 inputs, de peso y altura, que pasa a la funcion de calculo de imc
    
    Parámetros:
    peso (int o float): Primer número.
    altura (int o float): Segundo número.

    Retorna:
    int o float: cada variable separada.
    """
    peso = float(input('Cuánto pesas en kilos (usa punto si hay decimales): '))
    altura = float(input('Cuánto mides en metros (usa punto si hay decimales): '))
    return peso, altura

def calcular_imc(peso, altura):
    """
    recibe 2 inputs: de peso y altura y hace el calculo de imc
    
    Parámetros:
    peso (int o float): Primer número.
    altura (int o float): Segundo número.

    Retorna:
    int o float: el peso / por la altura ** 2.
    """
    return peso / altura ** 2

def separa_imc(imc):
    """
    recibe el imput del imc, redondea y retorna la categoria
    
    Parámetros:
    imc (int o float)

    Retorna:
    String: la categoria del imc recibido.
    """
    imc_redondeado = round(imc, 2)
    if imc < 18.5:
        categoria_imc = "Bajo Peso (menos de 18,5)"
    elif imc < 25:
        categoria_imc = "Peso Normal (entre 18,5 y 24,9)"
    elif imc < 30:
        categoria_imc = "Sobrepeso (entre 25 y 29,9)"
    else:
        categoria_imc = "Obesidad (más de 30)"
    
    return categoria_imc


def print_imc(imc, categoria_imc):
    """
    recibe 2 inputs: imc y categoria
    
    Parámetros:
    imc (int o float): Primer número.
    categoria (string): Texto con la categoria.

    Printa el resultado.
    """
    print(f"Tu IMC és {imc}, lo que se considera {categoria_imc}")


def contesta_imc():
    """
    funcion principal que llama todas las otras
    """
    peso, altura = pedir_datos()
    imc = calcular_imc(peso, altura)
    categoria_imc = separa_imc(imc)
    print_imc(imc, categoria_imc)

contesta_imc()
```

Convertidor de temperatures.

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una (cualquiera) temperatura retorni, com a mínim, dues conversions.

```python
def pedir_datos_temperatura():
    valor = float(input("Introduce la temperatura, solo el numero, y si tiene decimales escríbelo con punto (.): "))
    unidade_temperatura = input("Introduce la unidad de temperatura siendo letras minusculas. c para celcius, f para farenheit, k para kelvin, r para rankine y re para reamur")
    return valor, unidade_temperatura

def conversiones(valor, unidade_temperatura):
    diccionario_unidades_temperatura = {
        'c': '°C',
        'f': '°F',
        'k': 'K',
        'r': '°R',
        're': '°Re'
    }

    diccionario_conversiones = {
    # Celsius a otras
    'ctof': lambda c: c * 9/5 + 32, #lambda és una funcion sin nombre, de una sola linea. argumento y expresion adelante.
    'ctok': lambda c: c + 273.15,
    'ctor': lambda c: (c + 273.15) * 9/5,
    'ctore': lambda c: c * 4/5,

    # Fahrenheit a otras
    'ftoc': lambda f: (f - 32) * 5/9,
    'ftok': lambda f: (f - 32) * 5/9 + 273.15,
    'ftor': lambda f: f + 459.67,
    'ftore': lambda f: (f - 32) * 4/9,

    # Kelvin a otras
    'ktoc': lambda k: k - 273.15,
    'ktof': lambda k: (k - 273.15) * 9/5 + 32,
    'ktor': lambda k: k * 9/5,
    'ktore': lambda k: (k - 273.15) * 4/5,

    # Rankine a otras
    'rtoc': lambda r: (r - 491.67) * 5/9,
    'rtof': lambda r: r - 459.67,
    'rtok': lambda r: r * 5/9,
    'rtore': lambda r: ((r - 491.67) * 5/9) * 4/5,

    # Réaumur a otras
    'retoc': lambda re: re * 5/4,
    'retof': lambda re: re * 9/4 + 32,
    'retok': lambda re: re * 5/4 + 273.15,
    'retor': lambda re: (re * 5/4 + 273.15) * 9/5
    }
    

    for clave in diccionario_conversiones.keys():
        unidades_splitted = clave.split('to')
        if unidade_temperatura == unidades_splitted[0]:
            print(f"{valor}{diccionario_unidades_temperatura[unidades_splitted[0]]} equivale a {diccionario_conversiones[clave](valor):.2f}{diccionario_unidades_temperatura[unidades_splitted[1]]}")

def calcula_temperatura():
    valor, unidade_temperatura = pedir_datos_temperatura()
    conversiones(valor, unidade_temperatura)

calcula_temperatura()
```

Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
from collections import Counter
import re

def clean(text):
    clean_text = re.findall(r'\b\w+\b', text.lower())
    return clean_text

def count(clean_text):
    count = Counter(clean_text)
    return count

def pedir_datos_repeticion_palabras():
    text = input('haz un imput de un texto que te diré cuantas veces se repiten las palabras: ')
    return text

def funcio_principal_text():
    text = pedir_datos_repeticion_palabras()
    clean_text = clean(text)
    conteo = count(clean_text)
    print(conteo)

funcio_principal_text()
```

Diccionari invers.

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.

```python
dicionariconrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 2"
}

def reversio_diccionari(dicionariconrep):
    reversedictionary = {}

    for key, value in dicionariconrep.items():
        if value in reversedictionary:
            print(f"se repete! hemos hecho hasta aqui")
            return reversedictionary
        reversedictionary[value] = key

    return reversedictionary

def imprimir(reverse_dictionary):
    if reverse_dictionary is None:
        print(f"No se puede seguir invertiendo el diccionario porque hay valores que se repiten. Hicimos hasta aqui: {reverse_dictionary}") 
    else:
        print(reverse_dictionary)

def funcio_principal_reversedic():
    reverse_dicitionary = reversio_diccionari(dicionariconrep)
    imprimir(reverse_dicitionary)

funcio_principal_reversedic()
```

Diccionari invers amb duplicats

Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
dicionariconrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 2",
}

def reversio_diccionari(dicionariconrep):
    reversedictionaryduplicat = {}

    for key, value in dicionariconrep.items():
        if value in reversedictionaryduplicat:
            if not isinstance (reversedictionaryduplicat[value], list):
                reversedictionaryduplicat[value] = [reversedictionaryduplicat[value]]
            reversedictionaryduplicat[value].append(key)
        else:
            reversedictionaryduplicat[value] = key

    return reversedictionaryduplicat

def funcio_principal_reversedic(dicionariconrep):
    reversedictionaryduplicat = reversio_diccionari(dicionariconrep)
    print(reversedictionaryduplicat)

funcio_principal_reversedic(dicionariconrep)
```

Conversió de tipus de dades

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
# mirando en los elementos dentro de las listas presentes para la creacion de solo 2 otras listas
lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']] #lista de ejemplo

#creacion de cada una de las listas. 
listafloat = [] #en esta entrarán los elementos que se pueden convertir en float
listanofloat = [] #en esta entrarán los elementos no convertibles a float.

def try_convert_float(val): #funcion que intenta convertir a float
    try: #funcao que tenta converter, se funciona retorna o valor em float perfeito
        return float(val)
    except (ValueError, TypeError): #se da algum error sea de value ou de type, retorna NONE
        return None

def process_item(lista): #funcion de procesar el elemento
    for item in lista:
        if isinstance(item, (list, tuple)): #si es del tipo lista o tupla
            process_item(item)  #pasa por la funcion otra vez pero ya con los elementos abiertos y no dentro de las tuplas/listas
        else:
            f = try_convert_float(item) #lleva a la funcion de intentar convertir en float
            print(f) #para ver que elementos que nao sao convertibles retornan "None" efectivamente
            if f is None:
                listanofloat.append(item) #append item en NO FLOAT LIST
            else:
                listafloat.append(item) #append item en FLOAT list

    return listanofloat, listafloat

def funcio_principal_flotants():
    process_item(lista)
    nofloat = listanofloat
    float = listafloat
    print("Convertibles a float:", float)
    print("No convertibles:", nofloat)

funcio_principal_flotants()
```
