Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:

La funció ha de classificar el resultat en les seves respectives categories

```python
def pedir_datos():
    peso = float(input('Cuánto pesas en kilos (usa punto si hay decimales): '))
    altura = float(input('Cuánto mides en metros (usa punto si hay decimales): '))
    return peso, altura

def calcular_imc(peso, altura):
    return peso / altura ** 2

def mostrar_resultado(imc):
    imc_redondeado = round(imc, 2)
    if imc < 18.5:
        print(f"Tu IMC es: {imc_redondeado}, lo que se considera Bajo Peso (menos de 18,5)")
    elif imc < 25:
        print(f"Tu IMC es: {imc_redondeado}, lo que se considera Peso Normal (entre 18,5 y 24,9)")
    elif imc < 30:
        print(f"Tu IMC es: {imc_redondeado}, lo que se considera Sobrepeso (entre 25 y 29,9)")
    else:
        print(f"Tu IMC es: {imc_redondeado}, lo que se considera Obesidad (más de 30)")

def main():
    peso, altura = pedir_datos()
    imc = calcular_imc(peso, altura)
    mostrar_resultado(imc)

main()
```

Convertidor de temperatures.

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una (cualquiera) temperatura retorni, com a mínim, dues conversions.

```python
def desde_celsius(valor):
    f = valor * 9/5 + 32
    k = valor + 273.15
    ra = (valor + 273.15) * 9/5
    re = valor * 4/5
    print(f"Si {valor} °C es igual a:")
    print(f"  {round(f, 2)} °F")
    print(f"  {round(k, 2)} K")
    print(f"  {round(ra, 2)} °Ra")
    print(f"  {round(re, 2)} °Re")

def desde_fahrenheit(valor):
    c = (valor - 32) * 5/9
    k = c + 273.15
    ra = (c + 273.15) * 9/5
    re = c * 4/5
    print(f"Si {valor} °F es igual a:")
    print(f"  {round(c, 2)} °C")
    print(f"  {round(k, 2)} K")
    print(f"  {round(ra, 2)} °Ra")
    print(f"  {round(re, 2)} °Re")

def desde_kelvin(valor):
    c = valor - 273.15
    f = c * 9/5 + 32
    ra = valor * 9/5
    re = c * 4/5
    print(f"Si {valor} K es igual a:")
    print(f"  {round(c, 2)} °C")
    print(f"  {round(f, 2)} °F")
    print(f"  {round(ra, 2)} °Ra")
    print(f"  {round(re, 2)} °Re")

def desde_rankine(valor):
    c = (valor - 491.67) * 5/9
    f = valor - 459.67
    k = valor * 5/9
    re = c * 4/5
    print(f"Si {valor} °Ra es igual a:")
    print(f"  {round(c, 2)} °C")
    print(f"  {round(f, 2)} °F")
    print(f"  {round(k, 2)} K")
    print(f"  {round(re, 2)} °Re")

def desde_reaumur(valor):
    c = valor * 5/4
    f = c * 9/5 + 32
    k = c + 273.15
    ra = (c + 273.15) * 9/5
    print(f"Si {valor} °Re es igual a:")
    print(f"  {round(c, 2)} °C")
    print(f"  {round(f, 2)} °F")
    print(f"  {round(k, 2)} K")
    print(f"  {round(ra, 2)} °Ra")

def convertir_todas(valor):
    desde_celsius(valor)
    desde_fahrenheit(valor)
    desde_kelvin(valor)
    desde_rankine(valor)
    desde_reaumur(valor)

# Entrada del usuario
valor = float(input("Introduce la temperatura: "))
convertir_todas(valor)
```

Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
import re # Importar modulo RE para tratar texto

def count_palabras(text):
    text = text.lower()  # Convierte mayusculas y minusculas
    texto_limpio = re.sub(r"[(),;.!¡]", "", text) #limpiar texto por si hay caracteres raros
    palabras = texto_limpio.split() # split el texto en lista de palabras
    dicc_count = {} # Crear diccionario para guardar el count de palabras
    
    for cadapalabra in palabras:
        if cadapalabra in dicc_count:
            dicc_count[cadapalabra] += 1 # Si está en el dicc se añade aumenta en 1 el count
        else:
            dicc_count[cadapalabra] = 1 # Si no está en el dicc se añade y pone el count 1
    
    for cadapalabra, count in dicc_count.items():
        print(f"{cadapalabra}: {count}")


text = input('haz un imput de un texto que te diré cuantas veces se repiten las palabras: ')

count_palabras(text)
```

Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
dicionariconrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 3",
    "key 4" : "valor 1",
    "key 5" : "valor 1",
}

reversedictionary = {}

for key, value in dicionariconrep.items():
    if value in reversedictionary: #si el valor ya existe, entra en la conversion a lista
        if not isinstance(reversedictionary[value], list): # si no es del tipo lista, entra para convertir en lista
            reversedictionary[value] = [reversedictionary[value]]  # [] hace con que se convierta en lista
        reversedictionary[value].append(key) # si ya existe y ya es lista no entra en el bucle anterior y solo append el valor
    else:
        reversedictionary[value] = key # si no existia antes, solo atribuye el valor, sin atribuir a lista


print(reversedictionary)
```

Conversió de tipus de dades

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
# mirando en los elementos dentro de las listas presentes para la creacion de solo 2 otras listas

lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']] #lista de ejemplo

#creacion de cada una de las listas. 
listafloat = [] #en esta entrarán los elementos que se pueden convertir en float
listanofloat = [] #en esta entrarán los elementos no convertibles a float.

def try_convert_float(val): #funcion que intenta convertir a float
    try: #funcao que tenta converter, se funciona retorna o valor em float perfeito
        return float(val)
    except (ValueError, TypeError): #se da algum error sea de value ou de type, retorna NONE
        return None

def process_item(item): #funcion de procesar el elemento
    if isinstance(item, (list, tuple)): #si es del tipo lista o tupla
        for subitem in item: #entrar en cada subitem de la lista o tupla
            process_item(subitem)  #pasa por la funcion otra vez pero ya con los elementos abiertos y no dentro de las tuplas/listas
    else:
        f = try_convert_float(item) #lleva a la funcion de intentar convertir en float
        print(f) #para ver que elementos que nao sao convertibles retornan "None" efectivamente
        if f is None:
            listanofloat.append(item) #append item en no float list
        else:
            listafloat.append(item) #append item en FLOAT list

for x in lista: # para cada elemento dentro de la lista,
    process_item(x) #pasar por la funcion de procesar el elemento

print("Convertibles a float:", listafloat)
print("No convertibles:", listanofloat)
```

Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
import re # Importar modulo RE (regular expresions) para tratar texto
from collections import defaultdict # Importar default dict para atualizacion automatica (para crear automaticamente un valor 0 cuando se crea key-valor)
from pprint import pprint  # Importar pretty print

# Importar archivo y meter el archivo en una variable y conferir si está bien, printando en tipo y el proprio contenido
with open("tu_me_quieres_blanca.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
    print(contenido)
    print(type(contenido))

#Limpiar la variable contenido usando en modulo RE de python que he importado que ayuda a hacer manipulaciones en textos
texto_limpio = re.sub(r"[(),;.!¡]", "", contenido)     # quita signos. la R de raw string se utiliza para que los saltos de linea no se hagan pero si que se escriban \n literalmente, así se puede quitar luego
texto_limpio = re.sub(r"\s+", " ", texto_limpio)       # reemplaza saltos, tabs y espacios (no solo saltos como seria con \n) múltiples (por eso el +) por 1 solo espacio 

# crear una lista a partir del texto tractado, splitting cada palabra como un item
lista_palabras = texto_limpio.split()
print(lista_palabras)

# crear el subdiccionario con conteo automático donde el valor atribuido es 0 si hay una nueva key
def crear_subdiccionario():
    return defaultdict(int)

# diccionario principal, que creará subdiccionarios automáticamente buscando el DEF. también se podria hacer con "lambda: defaultdict(int)"
diccionario = defaultdict(crear_subdiccionario)

# atribuicion al subdiccionario de LETRAS y actualizacion de valor de cada palabra
for palabra in lista_palabras: #percorre la lista
    letra = palabra[0].lower() # pon en lowercase para comparar primera letra y atribui esta letra a la variable LETRA
    diccionario[letra][palabra] += 1 #accede el key LETRA y luego el key PALABRA y entonces al valor de PALABRA suma 1 al valor existente previamente

# pprint printa pretty para que se vea mejor
pprint(diccionario)

```
