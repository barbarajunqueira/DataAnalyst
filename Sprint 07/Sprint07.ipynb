Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:

La funció ha de classificar el resultat en les seves respectives categories

```python
peso = float(input('Cuánto pesas en quilos (número con punto si quieres poner gramos): '))
altura = float(input('Cuánto mides en metros (número con punto si necesitas poner centímetros): '))

imc = peso / altura ** 2

if imc < 18.5:
    print ("Tu IMC es: ", round(imc,2), "lo que se considera Bajo Peso (Menos de 18,5)")
elif 18.5 <= imc <25:
    print ("Tu IMC es: ", round(imc,2), "lo que se considera Peso Normal (Entre 18,5-24,9)")
elif 25<= imc < 30:
    print ("Tu IMC es: ", round(imc,2), "lo que se considera Sobrepeso (Entre 25-29,9)")
elif imc > 30:
    print ("Tu IMC es: ", round(imc,2), "lo que se considera Obesidad (Más que 30)")
```

Convertidor de temperatures.

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.

```python
c = float(input('Insira un valor de temperatura en Celcius (°C) para saber la equivalencia en Fahrenheit (°F) i Kelvin (K)'))

f = 9/5*c+32
k = c + 273.15

print ("Has inserido ", round(c,2),"°C y eso equivale a ",f, "°F (Fahrenheit) y ",k, "K (Kelvin)")
```

Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
def count_palabras(text):
    text = text.lower()  # Convierte mayusculas y minusculas
    palabras = text.split() # Substituïm caràcters especials per espais i dividim el text en una llista de paraules
    dicc_count = {} # Crear diccionario para guardar el count de palabras
    
    for cadapalabra in palabras:
        if cadapalabra in dicc_count:
            dicc_count[cadapalabra] += 1 # Si está en el dicc se añade aumenta en 1 el contage
        else:
            dicc_count[cadapalabra] = 1 # Si no está en el dicc se añade y pone el count 1
    
    for cadapalabra, count in dicc_count.items():
        print(f"{cadapalabra}: {count}")


text = input('haz un imput de un texto que te diré cuantas veces se repiten las palabras: ')

count_palabras(text)
```

Diccionari invers.

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.

```python
dicionarisinrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 3",
    "key 4" : "valor 4",
}

dicionariconrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 3",
    "key 4" : "valor 1",
}

reversedictionary = {}

for key, value in dicionariconrep.items():
    if value in reversedictionary:
        print(f"Valor {value} aparece más de una vez y por eso no se puede invertir el diccionario")
        break
    reversedictionary[value] = key


print(reversedictionary)
```

Diccionari invers amb duplicats

Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
dicionarisinrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 3",
    "key 4" : "valor 4",
}

dicionariconrep = {
    "key 1" : "valor 1",
    "key 2" : "valor 2",
    "key 3" : "valor 3",
    "key 4" : "valor 1",
}

reversedictionary = {}

for key, value in dicionariconrep.items():
    if value in reversedictionary:
        if not isinstance(reversedictionary[value], list): # si no es una lista
            reversedictionary[value] = [reversedictionary[value]]  # [] hace con que se convierta en lista
        reversedictionary[value].append(key)
    else:
        reversedictionary[value] = key


print(reversedictionary)
```

Conversió de tipus de dades

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
# considerando que la lista es un tipo no int, string o float, dejamos como lista, por si no se quiere desarmar la lista

lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

listafloat = []
listanofloat = []

def try_convert_float(val):
    try:
        return float(val)
    except (ValueError, TypeError):
        return None

for x in lista:
    # Si es str o numérico directo
    if isinstance(x, (int, float, str)):
        f = try_convert_float(x)
        if f is not None:
            listafloat.append(f)
        else:
            listanofloat.append(x)
    else:
        # Si es tupla o lista, no intentamos convertir directamente
        listanofloat.append(x)

print("Convertibles a float:", listafloat)
print("No convertibles:", listanofloat)
```

```python
# mirando en los elementos dentro de las listas presentes para la creacion de solo 2 otras listas

lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

listafloat = []
listanofloat = []

def try_convert_float(val):
    try:
        return float(val)
    except (ValueError, TypeError):
        return None

def process_item(item):
    if isinstance(item, (list, tuple)):
        for subitem in item:
            process_item(subitem)  # Recursión para manejar estructuras anidadas
    else:
        f = try_convert_float(item)
        if f is not None:
            listafloat.append(f)
        else:
            listanofloat.append(item)

for x in lista:
    process_item(x)

print("Convertibles a float:", listafloat)
print("No convertibles:", listanofloat)
```

Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
import re # Importar modulo RE para tratar texto
from collections import defaultdict # Importar default dict para atualizacion automatica
from pprint import pprint  # Importar pretty print

# Importar archivo y meter el archivo en una variable y conferir si está bien, printando en tipo y el proprio contenido
with open("tu_me_quieres_blanca.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
    print(contenido)
    print(type(contenido))

#Limpiar la variavle contenido usando en modulo RE de python que he importado que ayuda a hacer manipulaciones en textos
texto_limpio = re.sub(r"[(),;.!¡]", "", contenido)     # quita signos
texto_limpio = re.sub(r"\s+", " ", texto_limpio)       # reemplaza saltos, tabs y espacios múltiples por 1 solo espacio

# crear una lista a partir del texto tractado, splitting cada palabra como un item
lista_palabras = texto_limpio.split()
print(lista_palabras)

# crear el subdiccionario con conteo automático donde el valor atribuido es 0 si hay una nueva key
def crear_subdiccionario():
    return defaultdict(int)

# diccionario principal, que creará subdiccionarios automáticamente buscando el DEF. también se podria hacer con "lambda: defaultdict(int)"
diccionario = defaultdict(crear_subdiccionario)

# atribuicion al subdiccionario de LETRAS y actualizacion de valor de cada palabra
for palabra in lista_palabras: #percorre la lista
    letra = palabra[0].lower() # pon en lowercase para comparar primera letra y atribui esta letra a la variable LETRA
    diccionario[letra][palabra] += 1 #accede el key LETRA y luego el key PALABRA y entonces al valor de PALABRA suma 1 al valor existente previamente

# pprint printa pretty para que se vea mejor
pprint(diccionario)

```
